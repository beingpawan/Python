# Boolean Mask
## printing those values whose Marks came true

```python
Scores = np.array([64,67,78,95,46])
Marks=Scores>=65
print(Marks)
print(Scores[Scores>=65].mean())
print(Scores[(Scores>=60) & (Scores<=90)])
print(len(Scores>=60))
```
